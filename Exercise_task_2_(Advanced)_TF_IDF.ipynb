```python
import re
import math
from collections import defaultdict, Counter

# --------------------------------------------------
# 1. Downsample dataset
# --------------------------------------------------
texts_small = texts[:1000]
N = len(texts_small)

print("Number of documents used:", N)

```

```python
# --------------------------------------------------
# 2. Tokenization
# Replace this function with your subword tokenizer
# from the tokenization notebook if needed
# --------------------------------------------------
def tokenize_text(text):
    text = text.lower()
    tokens = re.findall(r"\b\w+\b", text)
    return tokens
```

```python
# --------------------------------------------------
# 3. Compute document frequencies
# --------------------------------------------------
df = defaultdict(int)

for doc in texts_small:
    tokens = tokenize_text(doc)
    unique_tokens = set(tokens)
    for token in unique_tokens:
        df[token] += 1

print("Number of unique tokens:", len(df))
```

```python
# --------------------------------------------------
# 4. Compute IDF
# --------------------------------------------------
idf = {}
for token, doc_freq in df.items():
    idf[token] = math.log((N + 1) / (doc_freq + 1)) + 1
```

```python

# --------------------------------------------------
# 5. Show low and high IDF tokens
# --------------------------------------------------
sorted_by_idf = sorted(idf.items(), key=lambda x: x[1])

print("\n10 tokens with the lowest IDF values:")
for token, score in sorted_by_idf[:10]:
    print(token, "->", round(score, 4))

print("\n10 tokens with the highest IDF values:")
for token, score in sorted_by_idf[-10:]:
    print(token, "->", round(score, 4))
```

```python
# --------------------------------------------------
# 6. TF function
# --------------------------------------------------
def compute_tf(tokens):
    tf = {}
    token_counts = Counter(tokens)
    total_tokens = len(tokens)

    for token, count in token_counts.items():
        tf[token] = count / total_tokens

    return tf
```

```python
# --------------------------------------------------
# 7. Print top TF-IDF tokens for one document
# --------------------------------------------------
def print_top_tfidf_words(document, idf_dict, top_k=10):
    tokens = tokenize_text(document)
    tf = compute_tf(tokens)

    tfidf_scores = {}
    for token in tf:
        if token in idf_dict:
            tfidf_scores[token] = tf[token] * idf_dict[token]

    sorted_tfidf = sorted(tfidf_scores.items(), key=lambda x: x[1], reverse=True)

    print(f"\nTop {top_k} tokens with highest TF-IDF values:\n")
    for token, score in sorted_tfidf[:top_k]:
        print(token, "->", round(score, 6))
```

```python
# --------------------------------------------------
# 8. Test on one document
# --------------------------------------------------
print("\nFirst document preview:\n")
print(texts_small[0][:500])

print_top_tfidf_words(texts_small[0], idf, top_k=10)
```
